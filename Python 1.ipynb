```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Supermart Grocery Sales - Retail Analytics Dataset.csv")
```

```python
df
```

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

city_sales.head(10).plot(kind="bar")

plt.title("Top 10 Cities by Sales")
plt.ylabel("Sales")
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

category_profit.plot(kind="bar")

plt.title("Profit by Category")
plt.ylabel("Profit")
plt.show()
```

```python
subcategory_sales = (
    df.groupby("Sub Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

subcategory_sales.head(10).plot(kind="barh")

plt.title("Top Selling Product Categories")
plt.xlabel("Sales")
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="pie", autopct="%1.1f%%")

plt.title("Sales by Region")
plt.ylabel("")
plt.show()
```

```python
#The analysis shows that sales are distributed quite evenly across the top cities, with Kanyakumari generating the highest sales. 
#The most profitable product categories are Snacks and Eggs, Meat & Fish, while Health Drinks and Soft Drinks generate the highest sales among product
#sub-categories. The West region contributes the largest share of total sales, making it the strongest-performing region. Based on these findings, the
#company should focus on high-performing regions and product categories to maximize revenue and profit.
```
